```python
import numpy as np

# Machine epsilon for float32 is 2^-23 (~1.19e-07)
eps_32 = np.finfo(np.float32).eps
print(f"Float32 Epsilon: {eps_32}")

# Choose a value smaller than eps_32 that cannot be distinctly represented
tiny_delta = np.float32(1e-8) 

# Assign 1.0 and 1.0 + tiny_delta as float32
a = np.float32(1.0)
b = np.float32(1.0 + tiny_delta)

# Because 1e-8 is less than the gap between representable float32 numbers at 1.0, 
# 'b' gets rounded down to exactly 1.0.
print("Does 1.0 + 1e-8 equal 1.0 in float32?", a == b)  # True

# Subtracting them yields 0.0 instead of the expected 1e-8 due to underflow/rounding
result = b - a
print(f"Subtraction result: {result}")  # 0.0
```
