```python
import numpy as np

float32_precision_magnitude= np.full((1,200),fill_value=1e-6,dtype=np.float32)
spectral_complexity_height_error = np.linalg.norm(float32_precision_magnitude, axis=1, keepdims=True)
print(spectral_complexity_height_error)


float64_precision_magnitude= np.full((1,200),fill_value=1e-15,dtype=np.float64)
spectral_complexity_height_error = np.linalg.norm(float64_precision_magnitude, axis=1, keepdims=True)
print(spectral_complexity_height_error)
```
